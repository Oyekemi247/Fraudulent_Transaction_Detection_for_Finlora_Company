###DATA CLEANING AND PREPARATION
converting datatypes of time stampe to datetime

```python
customer_transaction_data['timestamp']= pd.to_datetime(customer_transaction_data['timestamp'],errors='coerce')
#converting data type of income_usd to float
customer_transaction_data['amount_src']= pd.to_numeric(customer_transaction_data['amount_src'],errors='coerce')
```

```python
# confirming the data type again
customer_transaction_data.info()
```

To standardise data types, pd.to_datetime() and pd.to_numeric() were executed on timestamp and amount_src, utilising the errors='coerce' parameter to set invalid inputs to null. This explicit coercion surfaced underlying data corruption that had previously escaped detection as object types. Consequently, missing record counts rose from 29 to 61 for timestamp and from 0 to 4 for amount_src. These coerced missing values were systematically addressed during the broader data imputation and cleaning phase.

```python
#checking for missing data again
customer_transaction_data.isnull().sum()
```


•	amount_usd (305 missing): Imputed deterministically by multiplying amount_src by the corresponding currency exchange rate.

•	fee (295 missing): Imputed using conditional median strategy—filled by the median fee grouped by channel. Remaining missing values filled using overall median fee.

•	ip_country (301 missing): Imputed using the user's corresponding home_country.

•	kyc_tier (300 missing): Imputed using the global mode (most frequent value).

•	device_trust_score (295 missing): Imputed using conditional median strategy grouped by kyc_tier and new_device. Remaining missing values filled using overall median device_trust_score.


 timestamp variable which has 61 missing values and amount_src with 4 missing values will be dropped using the dropna method because the number of missing values in both vaiables is very small compared to the values on our dataset. The ip_address variable has missing values of 305 cannot be filled with any value as each ip_address is a unique number. It will be dropped using the dropna method.

```python
#calculating the exchange rate per currency
exchange_rates = customer_transaction_data[customer_transaction_data['amount_usd'].notna()].groupby('source_currency').apply(
    lambda x: (x['amount_usd'] / x ['amount_src']).mean()
).to_dict()
exchange_rates
```

```python
#To fill the missing values of the amount_usd
customer_transaction_data['amount_usd']= customer_transaction_data.apply(
    lambda row: row['amount_usd'] if pd.notna(row['amount_usd']) else row['amount_src'] * exchange_rates.get(row['source_currency'],1),
    axis=1
)
```

```python
# filling time missing vlues of the fee variable
if 'fee' in customer_transaction_data.columns:
    if 'channel' in customer_transaction_data:
        customer_transaction_data['fee'] = customer_transaction_data['fee'].fillna(customer_transaction_data['fee'].median())
```

```python
# filling the missing valuesof ip_country variable
if {'ip_country', 'home_country'}.issubset(customer_transaction_data.columns):
    customer_transaction_data['ip_country'] = customer_transaction_data['ip_country'].fillna(customer_transaction_data['home_country'])

    
```

```python
# to fill the missing values for device_trust_score variable
if 'device_trust_score' in customer_transaction_data.columns:
   if {'if new_device', 'kyc_tier'}.issubset(customer_transaction_data.columns):
    customer_transaction_data['device_trust_score']=customer_transaction_data.groupby(['new_device','kyc_tier',])['device_trust_score'].tramsform(lambda x: x.fillna(x.median()))
    customer_transaction_data['device_trust_score']=customer_transaction_data['device_trust_score'].fillna(customer_transaction_data['device_trust_score'].median())
```

```python
#To drop the timestamp, amount_src and ip_address missing values
customer_transaction_data.dropna(subset=['timestamp', 'amount_src','ip_address'], inplace=True)
```
